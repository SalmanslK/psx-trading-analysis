```python
import requests
from bs4 import BeautifulSoup
import re
import time
```

```python
def scrape_and_calculate_average(url):
    # Make an HTTP request to the specified URL
    response = requests.get(url)

    # Check if the request was successful (status code 200)
    if response.status_code == 200:
        # Parse the HTML content of the page using BeautifulSoup
        soup = BeautifulSoup(response.content, 'html.parser')

        # Find the first 15 divs with the class 'price-details'
        price_details_divs = soup.find_all('div', class_='price-details')[:15]

        #print("No of Prices found:")
        #print(len(price_details_divs))
        
        # Extract the numbers from each div and store them in a list
        numbers = []
        for div in price_details_divs:
            text = div.get_text()
            # Use regular expression to extract numeric part
            numeric_part = re.search(r'\d+\.\d+', text)
            if numeric_part:
                numbers.append(float(numeric_part.group()))
                
        #print(numbers)

        # Calculate the average of the extracted numbers
        if numbers:
            average = sum(numbers) / len(numbers)
            print(f"Average Price: {average:.2f}")
        else:
            print("No valid price details found.")
    else:
        print(f"Failed to retrieve the page. Status code: {response.status_code}")
```

```python
# Replace 'your_url_here' with the actual URL you want to scrape
url_to_scrape = 'https://www.pakwheels.com/used-cars/search/-/mk_suzuki/md_alto/ct_karachi/yr_2010_2010/'
scrape_and_calculate_average(url_to_scrape)
```
